# 📒 Lekce 01 – Slovníky (Dictionaries)

Zápisník z kurzu [Úvod do programování: Python](https://kodim.cz/czechitas/uvod-do-progr-1) na Kódím.cz.

Slovník (`dict`) je datová struktura, která ukládá data jako **páry klíč–hodnota**.  
Na rozdíl od seznamu nepřistupujeme k hodnotám přes index (0, 1, 2...), ale přes **pojmenovaný klíč**.

```
{ klíč: hodnota, klíč: hodnota, ... }
```

Hodí se všude tam, kde chceme data pojmenovat – vysvědčení, kontakt, konfiguraci, tombola...

## 1. Vytvoření slovníku

````python
# Nejběžnější způsob – složené závorky
vysvedceni = {
    "Matematika": 1,
    "Biologie": 2,
    "Chemie": 1,
}
print(vysvedceni)
````

````python
# Pomocí funkce dict() – přehledné pro jednoduché klíče bez mezer
kontakt = dict(jmeno="Jana", telefon="777 123 456", mesto="Praha")
print(kontakt)
````

````python
# Ze seznamu dvojic (tuple) – hodí se když data přicházejí v téhle formě
dvojice = [("a", 1), ("b", 2), ("c", 3)]
pismena = dict(dvojice)
print(pismena)
````

## 2. Přístup k hodnotám

````python
vysvedceni = {"Matematika": 1, "Biologie": 2, "Chemie": 1}

# Přístup přes klíč – pozor, pokud klíč neexistuje, dostaneme KeyError
print(vysvedceni["Matematika"])

# Bezpečnější varianta – metoda get()
# Pokud klíč neexistuje, vrátí None (nebo námi zvolenou výchozí hodnotu)
print(vysvedceni.get("Fyzika"))                    # → None
print(vysvedceni.get("Fyzika", "předmět nenalezen"))  # → předmět nenalezen
````

````python
# Praktický příklad z cvičení – tombola
# get() se tady opravdu hodí, protože ne každý lístek vyhraje
tombola = {
    7: "Láhev vína",
    23: "Čokoládový dort",
    67: "Vyhlídkový let balónem",
}
listek = 23
vyhra = tombola.get(listek, "Bohužel nevyhráváš nic.")
print(vyhra)
````

## 3. Přidávání a úprava položek

````python
sales = {
    "Zkus mě chytit": 4165,
    "Vrah zavolá v deset": 5681,
    "Zločinný steh": 2565,
}

# Přidání nové položky – jednoduše přiřadíme nový klíč
sales["Noc, která mě zabila"] = 0
print(sales)

# Úprava existující hodnoty
sales["Vrah zavolá v deset"] += 100
print(f"Po úpravě: {sales['Vrah zavolá v deset']}")
````

## 4. Mazání položek

````python
slovnik = {"a": 1, "b": 2, "c": 3, "d": 4}

# del – smaže klíč (KeyError pokud neexistuje)
del slovnik["a"]
print("Po del:", slovnik)

# pop() – smaže A zároveň vrátí hodnotu; s výchozí hodnotou je bezpečný
hodnota = slovnik.pop("b")
print(f"Smazáno: {hodnota}, zbytek: {slovnik}")

chybejici = slovnik.pop("z", "klíč neexistuje")
print(chybejici)

# clear() – vymaže celý slovník najednou
slovnik.clear()
print("Po clear:", slovnik)
````

## 5. Procházení slovníku

Tady jsem si musel dát pozor – `for key in slovnik` prochází **jen klíče**.  
Pokud chci i hodnoty, musím použít `.items()` – to pak rozbalím do dvou proměnných.

````python
sales = {
    "Zkus mě chytit": 4165,
    "Vrah zavolá v deset": 5681,
    "Zločinný steh": 2565,
}

print("--- Pouze klíče ---")
for kniha in sales:
    print(kniha)

print("\n--- Pouze hodnoty ---")
for pocet in sales.values():
    print(pocet)

print("\n--- Klíče i hodnoty (nejčastěji používám) ---")
total = 0
for kniha, pocet in sales.items():
    print(f"  '{kniha}' → {pocet} výtisků")
    total += pocet

print(f"\nCelkem: {total} výtisků")
````

## 6. Seznam slovníků

Tohle mi ze začátku dělalo problémy – hodnota slovníku může být **cokoliv**, tedy i další slovník nebo seznam.  
Nejčastější případ v praxi: seznam záznamů (knihy, produkty, osoby...).

````python
# Příklad z kurzu – seznam knih, každá kniha je slovník
books = [
    {"title": "Zkus mě chytit",       "sold": 4165, "price": 347, "year": 2018},
    {"title": "Vrah zavolá v deset",  "sold": 5681, "price": 299, "year": 2019},
    {"title": "Zločinný steh",        "sold": 2565, "price": 369, "year": 2019},
    {"title": "Bidnici",              "sold": 3549, "price": 669, "year": 2018},
    {"title": "Šoupátka",             "sold": 6843, "price": 429, "year": 2020},
]

# Celkový počet prodaných knih
celkem_knih = 0
for kniha in books:
    celkem_knih += kniha["sold"]

print(f"Celkem prodáno: {celkem_knih} knih")

# Tržba za konkrétní rok
rok = 2019
trzba = 0
for kniha in books:
    if kniha["year"] == rok:
        trzba += kniha["sold"] * kniha["price"]

print(f"Tržba za rok {rok}: {trzba:,} Kč")
````

## 7. Užitečné metody slovníku

````python
zakladni = {"a": 1, "b": 2}
doplnek  = {"b": 99, "c": 3}

# update() – sloučí dva slovníky, existující klíče přepíše
zakladni.update(doplnek)
print("update():", zakladni)   # → {"a": 1, "b": 99, "c": 3}

# copy() – mělká kopie, změny v kopii neovlivní originál
original = {"x": 10, "y": 20}
kopie = original.copy()
kopie["x"] = 999
print("original:", original)   # → {"x": 10, "y": 20}  ← nezměněn
print("kopie:   ", kopie)      # → {"x": 999, "y": 20}

# setdefault() – nastaví výchozí hodnotu POUZE pokud klíč ještě neexistuje
nastaveni = {"barva": "modrá"}
nastaveni.setdefault("barva", "červená")    # klíč existuje → nepřepíše
nastaveni.setdefault("velikost", "střední") # klíč chybí   → přidá
print("setdefault():", nastaveni)
````

Práce se slovníkem vysvědčení – výpis klíčů, počet jedniček, průměr známek.

````python
import statistics

school_report = {
    "Český jazyk": 1,    "Anglický jazyk": 1, "Matematika": 1,
    "Přírodopis": 2,     "Dějepis": 1,        "Fyzika": 2,
    "Hudební výchova": 4, "Výtvarná výchova": 2, "Tělesná výchova": 3, "Chemie": 4,
}

print("Předměty:", list(school_report.keys()))
print(f"Počet jedniček: {list(school_report.values()).count(1)}")

# Průměr – nejkratší varianta s modulem statistics
print(f"Průměr (statistics): {statistics.mean(school_report.values()):.2f}")

# Průměr – ručně přes for cyklus (abych viděl jak to funguje uvnitř)
celkem, pocet = 0, 0
for predmet, znamka in school_report.items():
    celkem += znamka
    pocet  += 1
    if znamka == 1:
        print(f"  Jedničku má z: {predmet}")

print(f"Průměr (ručně):      {celkem / pocet:.2f}")
````
