## 1. Základní definice funkce

Funkce pojmenuje blok kódu, který chceme opakovaně používat.  
Název funkce píšeme **malými písmeny** (konvence Pythonu).

```python
# Nejjednodušší funkce – přijme dva parametry, vrátí výsledek
def mult(a, b):
    vysledek = a * b
    return vysledek

print(mult(3, 7))    # → 21
print(mult(10, 0.5)) # → 5.0
```

```python
# Type hints – nepovinné, ale pomáhají editorům s nápovědou
# Samotný Python typy nevynucuje, jde jen o dokumentaci záměru
def greet_user(name: str, age: int) -> str:
    return f"Ahoj {name}, je ti {age} let."

print(greet_user("Jana", 28))
```

## 2. Výchozí hodnoty parametrů

Parametr může mít výchozí hodnotu – pak je volitelný.  
Povinné parametry musí být vždy **před** těmi s výchozí hodnotou.

```python
# Hotel – cena pobytu, snídaně je volitelná
def total_price(persons: int, breakfast: bool = False) -> int:
    cena_noc     = 850 * persons
    cena_snidane = 125 * persons
    return cena_noc + (cena_snidane if breakfast else 0)

print(total_price(3))           # bez snídaně  → 2550
print(total_price(2, True))     # se snídaní   → 1950
print(total_price(2, breakfast=True))  # explicitní pojmenování parametru – přehlednější
```

```python
# Výpočet známky z bodů – druhý výchozí parametr
def vypocti_znamku(body: int, bonusove_body: int = 0) -> int:
    celkove_body = body + bonusove_body
    if celkove_body > 20:
        return 1
    return 5

print(vypocti_znamku(15))              # → 5
print(vypocti_znamku(15, 10))          # → 1  (15 + 10 = 25)
print(vypocti_znamku(15, bonusove_body=10))  # stejné, čitelnější zápis
```

## 3. Funkce vracející `True` / `False`

Funkce nemusí vracet číslo – může vracet jakýkoliv typ, včetně `bool`.  
Pak ji lze rovnou použít v podmínce `if`.

```python
# Měsíc z rodného čísla
# Formát: RRMMDD... – u žen je MM +50 (např. leden žena = 51)
def month_of_birth(rodne_cislo: str) -> int:
    mesic = int(rodne_cislo[2:4])
    if mesic >= 50:
        mesic -= 50   # ženy mají +50, odečteme
    return mesic

print(month_of_birth("8503159876"))   # → 3  (muž, narozen v březnu)
print(month_of_birth("8563159876"))   # → 13? → zkontroluj: 63-50=13 ... hmm
# Tady jsem si uvědomil: hodnoty 13+ by byly chyba vstupu – v reálu bychom validovali
```

## 4. Bonus – Ruleta

Zajímavý příklad na funkci s `range()` a podmíněným `return`.  
Elegantní je jednořádkový zápis s ternárním operátorem `x if podminka else y`.

```python
import random

# Ruleta: čísla 1–36 rozdělena do 3 řad (1,4,7... / 2,5,8... / 3,6,9...)
# Sázka na řadu → výhra = dvojnásobek, prohra = 0
def ruleta(cislo_rady: int, sazka: int) -> int:
    rady = [
        range(1, 37, 3),  # řada 1: 1, 4, 7, ...
        range(2, 37, 3),  # řada 2: 2, 5, 8, ...
        range(3, 37, 3),  # řada 3: 3, 6, 9, ...
    ]
    hozeno = random.randint(0, 36)
    print(f"  hozeno: {hozeno}")
    # Ternární operátor – elegantní zkrácení if/else na jeden řádek
    return 2 * sazka if hozeno in rady[cislo_rady - 1] else 0

# Simulace 5 kol
print("Simulace rulety (řada 2, sázka 100 Kč):")
for kolo in range(1, 6):
    vyhra = ruleta(2, 100)
    print(f"  kolo {kolo}: výhra = {vyhra} Kč")
```

## 🏠 Domácí úkol – SMS brána

Zadání: [ukol-03](https://github.com/andywaltlova/python-1-podzim-2022/blob/master/ukoly/ukol-03.md)

Dvě funkce: ověření tel. čísla + výpočet ceny SMS (3 Kč / 180 znaků).

```python
import math

def cert_number(cislo: str) -> bool:
    """Vrátí True pokud je číslo 9místné nebo 13místné (s předvolbou +420)."""
    cislo_bez_mezer = cislo.replace(" ", "")
    delka = len(cislo_bez_mezer)
    return delka == 9 or delka == 13

def cena_sms(zprava: str) -> int:
    """Cena SMS: 3 Kč za každých započatých 180 znaků."""
    return math.ceil(len(zprava) / 180) * 3

# Testovací volání (bez input(), aby notebook šel spustit)
testovaci_cisla = ["777123456", "+420 777 123 456", "12345", "00420777123456"]
for cislo in testovaci_cisla:
    platne = cert_number(cislo)
    print(f"  '{cislo}' → {'platné ✓' if platne else 'neplatné ✗'}")

print()
zpravy = ["Ahoj!", "A" * 180, "A" * 181, "A" * 360]
for z in zpravy:
    print(f"  {len(z)} znaků → {cena_sms(z)} Kč")
```
