## 1. Třída, konstruktor a atributy

Třída je šablona pro objekty. Název třídy píšeme **PascalCase** (každé slovo velkým).  
`__init__` se zavolá automaticky při vytvoření objektu – tam nastavujeme atributy přes `self`.

```python
class Jednorozec:
    def __init__(self, jmeno: str, barva: str, as_alive: bool = False):
        self.jmeno    = jmeno
        self.barva    = barva
        self.as_alive = as_alive

    def __str__(self):
        # __str__ definuje co se vypíše při print(objekt)
        return f"{self.jmeno} má {self.barva} barvu"

    def __eq__(self, other):
        # __eq__ definuje co znamená ==  (tady: stejné jméno = stejný jednorozec)
        return self.jmeno == other.jmeno

    def vypis_informace(self):
        stav = "naživu" if self.as_alive else "mrtev"
        print(f"{self.jmeno} má barvu {self.barva} a je {stav}.")

karel = Jednorozec("Karel", "duhovou")
lenka = Jednorozec("Lenka", "pruhovanou", as_alive=True)

print(karel)
karel.vypis_informace()
lenka.vypis_informace()

# Test __eq__
karel2 = Jednorozec("Karel", "růžovou")
print(f"karel == karel2: {karel == karel2}")   # True – stejné jméno
print(f"karel == lenka:  {karel == lenka}")    # False
```

## 2. Metody měnící stav objektu

Metody třídy mohou měnit atributy objektu – objekt si pamatuje svůj stav.

```python
class Kniha:
    def __init__(self, nazev: str, pocet_stran: int, cena: float):
        self.nazev       = nazev
        self.pocet_stran = pocet_stran
        self.cena        = cena

    def __str__(self):
        return f"'{self.nazev}' ({self.pocet_stran} str.) – {self.cena:.0f} Kč"

    def sleva(self, velikost_slevy: float = 0):
        """Sníží cenu o zadané procento (0–100)."""
        self.cena *= 1 - (velikost_slevy / 100)

kniha1 = Kniha("Stopař", 120, 390)
kniha2 = Kniha("Dukla", 50, 150)

print(kniha1)
kniha1.sleva(20)          # -20 %
print(f"Po slevě: {kniha1}")
print(kniha2)
```

## 3. Kompozice – třída uvnitř třídy

Objekt jedné třídy může být atributem druhé. Tady `Kucharka` obsahuje seznam objektů `Recept`.

```python
class Recept:
    def __init__(self, nazev: str, narocnost: int, url: str):
        self.nazev      = nazev
        self.narocnost  = narocnost
        self.url        = url
        self.vyzkouseno = False

    def __str__(self):
        stav = "✓ vyzkoušeno" if self.vyzkouseno else "– nevyzkoušeno"
        return f"{self.nazev} (náročnost {self.narocnost}/5) {stav} → {self.url}"

    def __repr__(self):
        # __repr__ pro výpis v seznamu; tady stačí delegovat na __str__
        return self.__str__()

    def zmen_narocnost(self, nova: int):
        self.narocnost = nova

    def zkusit(self):
        self.vyzkouseno = True


class Kucharka:
    def __init__(self, nazev: str, autor: str):
        self.nazev   = nazev
        self.autor   = autor
        self.recepty = []

    def __str__(self):
        hlavicka = f"📖 {self.nazev} – {self.autor} ({len(self.recepty)} receptů)"
        radky = [f"  {r}" for r in self.recepty]
        return "\n".join([hlavicka] + radky)

    def pridej_recept(self, recept: Recept):
        self.recepty.append(recept)

    def vyzkousene(self) -> list:
        return [r for r in self.recepty if r.vyzkouseno]


# --- Ukázka ---
r1 = Recept("Babovka", 1, "babovky.cz")
r2 = Recept("Štrůdl",  2, "strudly.eu")
r3 = Recept("Buchta",  5, "buchty.cz")

r1.zmen_narocnost(3)
r1.zkusit()
r3.zkusit()

kucharka = Kucharka("Domácí kuchařka", "Tom")
kucharka.pridej_recept(r1)
kucharka.pridej_recept(r2)
kucharka.pridej_recept(r3)

print(kucharka)
print(f"\nVyzkoušené recepty: {kucharka.vyzkousene()}")
```

## 🏠 Domácí úkol – Recept a Kucharka (rozšíření)

```python
# Třída Balik a Zamestnanec ze cvičení
class Balik:
    def __init__(self, adresa: str, hmotnost: float):
        self.adresa    = adresa
        self.hmotnost  = hmotnost
        self.doruceno  = False

    def doruc(self):
        self.doruceno = True

    def __str__(self):
        stav = "doručen" if self.doruceno else "nedoručen"
        return f"Balík na {self.adresa} ({self.hmotnost} kg) – {stav}"

b1 = Balik("Valašské Meziříčí", 2.5)
b2 = Balik("Praha", 0.8)
print(b1)
b1.doruc()
print(b1)
print(b2)
```
