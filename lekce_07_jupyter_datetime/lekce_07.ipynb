## 2. Modul `datetime`

`datetime` je standardní knihovna Pythonu pro práci s datem a časem.  
Importujeme konkrétní třídy: `datetime`, `date`, `timedelta`, `timezone`.

```python
from datetime import datetime, timedelta, date, timezone
datetime.now()
```

## 3. Vytvoření konkrétního datumu a formátování

`datetime(rok, měsíc, den, hodina, minuta, sekunda)` – vytvoří konkrétní okamžik.  
`strftime(formát)` – převede datetime na řetězec v daném formátu.

```python
start_apolla = datetime(1996, 7, 16, 14, 32)
start_apolla
```

```python
datum_narozeni = datetime(1989, 10, 8, 22, 18, 4)

# isoweekday() vrátí číslo dne: 1=Po, 7=Ne
print(f"Den v týdnu (číslo): {datum_narozeni.isoweekday()}")

# strftime – formátování do čitelného textu
print(datum_narozeni.strftime("%d.%m.%Y"))       # → 08.10.1989
print(datum_narozeni.strftime("%H:%M %I %p"))    # → 22:18 10 PM
```

## 4. Rozdíl dat a `timedelta`

Odečtením dvou `datetime` dostaneme `timedelta` – délku intervalu.  
`timedelta` lze také vytvořit přímo a přičítat/odečítat od data.

```python
from datetime import datetime, timedelta, date, timezone

apollo_start = datetime(1969, 7, 16, 14, 32)
apollo_start_US = apollo_start.strftime("%m/%d/%Y")
print(apollo_start_US)
```

```python
orbiter_start = datetime(2020, 2, 10, 5, 3)
#den_start_orbiter = orbiter_start.weekday()
den_start_orbiter = orbiter_start.strftime("%A")
print(den_start_orbiter)
```

```python
orbiter_start = datetime(2020, 2, 10, 5, 3)
cas = datetime.now()

rozdil = cas - orbiter_start
print(rozdil)
```

```python
cas_objednavky = datetime(2020, 11, 13, 19, 47)
doba_prevzeti = timedelta(minutes=8, seconds=35)
doba_pripravy = timedelta(minutes=30)
doba_dopravy = timedelta(minutes=25, seconds=30)
doba_celkove = doba_prevzeti + doba_pripravy + doba_dopravy

cas_doruceni = cas_objednavky + doba_celkove
print(cas_doruceni)
```
