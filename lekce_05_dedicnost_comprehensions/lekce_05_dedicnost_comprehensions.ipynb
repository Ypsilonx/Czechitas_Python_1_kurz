## 1. Dědičnost

Podřízená třída zdědí vše z nadřízené. `super().__init__()` zavolá konstruktor rodiče –  
nemusíme opisovat společné atributy znovu.

```python
class Zamestnanec:
    def __init__(self, jmeno: str, pozice: str):
        self.jmeno              = jmeno
        self.pozice             = pozice
        self.pocet_dni_dovolene = 25

    def __str__(self):
        return f"{self.jmeno} – {self.pozice} (dovolená: {self.pocet_dni_dovolene} dní)"

    def cerpani_dovolene(self, pocet_dni: int) -> str:
        if pocet_dni <= self.pocet_dni_dovolene:
            self.pocet_dni_dovolene -= pocet_dni
            return f"OK, zbývá {self.pocet_dni_dovolene} dní"
        return f"Nemáš dostatek dovolené (zbývá {self.pocet_dni_dovolene} dní)"


class Manazer(Zamestnanec):
    def __init__(self, jmeno: str, pozice: str, pocet_podrizenych: int):
        super().__init__(jmeno, pozice)          # sdílené atributy z Zamestnanec
        self.pocet_podrizenych  = pocet_podrizenych
        self.pocet_dni_dovolene = 30             # manažer má o 5 dní více

    def __str__(self):
        # super().__str__() – přebírám text rodiče a doplním svůj
        return super().__str__() + f", podřízených: {self.pocet_podrizenych}"


frantisek = Zamestnanec("František", "programátor")
arnost    = Manazer("Arnošt", "vedoucí týmu", 10)

print(frantisek)
print(arnost)
print(frantisek.cerpani_dovolene(5))
print(arnost.cerpani_dovolene(10))   # Manazer zdědil metodu z Zamestnanec
```

## 2. List comprehension

Zkrácený zápis pro vytvoření seznamu transformací nebo filtrů.  
Forma: `[výraz for prvek in zdroj]` nebo `[výraz for prvek in zdroj if podmínka]`

```python
cisla = [1.12, 4.51, 2.64, 13.1, 0.1]

# Klasický for vs. comprehension – stejný výsledek, kratší zápis
zdvojene_for = []
for c in cisla:
    zdvojene_for.append(c * 2)

zdvojene_comp = [c * 2 for c in cisla]
print("for loop:     ", zdvojene_for)
print("comprehension:", zdvojene_comp)

# Různé transformace
jmena = ["Roman", "Jan", "Miroslav", "Petr", "Gabriel"]
print([jmeno.upper()          for jmeno in jmena])   # velká písmena
print([len(jmeno)             for jmeno in jmena])   # délky
print([jmeno.lower() + "@email.cz" for jmeno in jmena])  # e-maily
```

```python
# Comprehension s podmínkou – filtrování
pisemky = [0, 2, 0, 1, 1, 3]

# Opravit jen nuly (dát +1 místo 0)
opravene = [z + 1 if z == 0 else z for z in pisemky]
print("Opravené známky:", opravene)

# Vyfiltrovat jen nuly
jen_nuly = [z for z in pisemky if z == 0]
print("Počet nul:", len(jen_nuly))
```

```python
# Comprehension na vnořeném seznamu – teploty (7 dní × 4 měření)
teploty = [
    [2.1, 5.2, 6.1, -0.1],
    [2.2, 4.8, 5.6, -1.0],
    [3.3, 6.5, 5.9,  1.2],
    [2.9, 5.6, 6.0,  0.0],
    [2.0, 4.6, 5.5, -1.2],
    [1.0, 3.2, 2.1, -2.0],
    [0.4, 2.7, 1.3, -2.8],
]

denni_prumery  = [round(sum(d) / len(d), 1) for d in teploty]
ranni_teploty  = [d[0]  for d in teploty]
nocni_teploty  = [d[-1] for d in teploty]

print("Denní průměry:", denni_prumery)
print("Ranní:        ", ranni_teploty)
print("Noční:        ", nocni_teploty)
print(f"Týdenní průměr: {sum(denni_prumery)/len(denni_prumery):.1f} °C")
```

## 🏠 Domácí úkol – Koronavirus (dědičnost) + státy (comprehension)

Zadání: [ukol-05](https://github.com/andywaltlova/python-1-podzim-2022/blob/master/ukoly/ukol-05.md)

```python
class Nemoc:
    def __init__(self, jmeno: str, inkubacni_doba: int, pocet_obeti: int, sireni: str):
        self.jmeno          = jmeno
        self.inkubacni_doba = inkubacni_doba
        self.pocet_obeti    = pocet_obeti
        self.sireni         = sireni

    def __str__(self):
        return f"Nemoc: {self.jmeno}"

    def zmen_pocet_obeti(self, pocet: int):
        self.pocet_obeti = pocet


class Koronavirus(Nemoc):
    def __init__(self, jmeno: str, varianta: list = None):
        super().__init__(jmeno, inkubacni_doba=7, pocet_obeti=0, sireni="vzduchem")
        self.varianta = varianta if varianta is not None else []

    def __str__(self):
        if self.varianta:
            varianty = ", ".join(self.varianta)
            return super().__str__() + f" (varianty: {varianty})"
        return super().__str__() + " (žádné varianty)"

    def pridej_variantu(self, varianta: str):
        self.varianta.append(varianta)


corona = Koronavirus("Coronavirus")
print(corona)
corona.zmen_pocet_obeti(5000)
corona.pridej_variantu("omikron")
corona.pridej_variantu("delta")
print(corona)
print(f"Šíření: {corona.sireni}, inkubace: {corona.inkubacni_doba} dní, oběti: {corona.pocet_obeti}")
```

```python
import json
from pathlib import Path

staty = json.loads(Path("../data/staty.json").read_text(encoding="utf-8"))

# Comprehension na datech – státy s populací > 100 mil. v Asii
velke_asijske = [
    s["name"] for s in staty
    if s.get("region") == "Asia" and s.get("population", 0) > 100_000_000
]
print("Asijské státy s populací > 100M:", velke_asijske)

# Dict comprehension – slovník {název: populace} pro Evropu
evropa = {
    s["name"]: s["population"]
    for s in staty
    if s.get("region") == "Europe"
}
nejlidnatejsi = max(evropa, key=evropa.get)
print(f"\nNejlidnatější stát v Evropě: {nejlidnatejsi} ({evropa[nejlidnatejsi]:,} obyvatel)")
```
