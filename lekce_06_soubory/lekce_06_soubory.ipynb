## 1. Čtení souboru – `with open()`

`with` zajistí, že se soubor automaticky zavře i při chybě – proto ho vždy používám místo `open()` + `close()`.

```python
from pathlib import Path

# Přečti výkaz hodin – každý řádek = počet hodin za den
# .strip() odstraní \n a mezery z okrajů každého řádku
with open("../data/vykaz.txt", encoding="utf-8") as f:
    hodiny = [int(radek.strip()) for radek in f.readlines()]

print("Hodiny:", hodiny)
print("Celkem:", sum(hodiny), "h")
```

```python
# Počítání slov v textu – .split() rozdělí řádek na slova
with open("../data/praha.txt", encoding="utf-8") as f:
    radky = f.readlines()

slova_per_radek = [len(radek.strip().split()) for radek in radky]
celkem_slov = sum(slova_per_radek)
print(f"Počet řádků: {len(radky)}")
print(f"Celkem slov: {celkem_slov}")
```

## 2. Zápis do souboru

`mode="w"` přepíše soubor, `mode="a"` přidá na konec.

```python
jmena = ["Roman", "Jana", "Radek", "Petra", "Vlasta"]

# Zapíšeme jména do souboru, každé na nový řádek
with open("../data/uzivatele.txt", mode="w", encoding="utf-8") as f:
    for jmeno in jmena:
        f.write(jmeno + "\n")

# Ověřovací čtení – přečteme co jsme zapsali
with open("../data/uzivatele.txt", encoding="utf-8") as f:
    obsah = f.read()
print(obsah)
```

## 3. CSV soubory – modul `csv`

Modul `csv` je lepší než ruční `.split(",")` – správně zvládne uvozovky a speciální znaky.

```python
import csv

PREVOD = {"1": "A", "2": "B", "3": "C", "4": "D", "5": "F"}

def replace_values(radek: list, prevod: dict) -> list:
    """Převede číselné známky na písmenkové dle slovníku."""
    return [prevod.get(hodnota, hodnota) for hodnota in radek]

header = []
rows   = []

with open("../data/znamky_za_semestr.csv", encoding="utf-8") as csvf:
    reader = csv.reader(csvf)
    for i, row in enumerate(reader):
        if i == 0:
            header = row
        else:
            rows.append(replace_values(row, PREVOD))

print(f"Žáků: {len(rows)}")
print("Záhlaví:", header)
print("Ukázka (první 3 žáci):")
for r in rows[:3]:
    print(" ", r)
```

```python
# Zápis upraveného CSV
vystupni_soubor = "../data/znamky_za_semestr_upravene.csv"

with open(vystupni_soubor, mode="w", encoding="utf-8", newline="") as csvf:
    writer = csv.writer(csvf)
    writer.writerow(header)
    writer.writerows(rows)

print(f"Uloženo do: {vystupni_soubor}")
```

## 🏠 Domácí úkol – výkaz mzdy

Zadání: [ukol-06](https://github.com/andywaltlova/python-1-podzim-2022/blob/master/ukoly/ukol-06.md)

Načti hodiny z `vykaz.txt`, vypočítej měsíční mzdy a výsledek ulož do nového souboru.

```python
hodinovka = 350  # Kč/h (místo input() pro notebook)

with open("../data/vykaz.txt", encoding="utf-8") as f:
    hodiny = [int(r.strip()) for r in f.readlines()]

mzdy = [h * hodinovka for h in hodiny]
prumerna_mzda = sum(mzdy) // len(mzdy)

print(f"Hodinová mzda: {hodinovka} Kč")
print(f"Měsíční mzdy:  {mzdy}")
print(f"Průměr:        {prumerna_mzda} Kč")

# Zápis výsledku
with open("../data/vykaz1.txt", mode="w", encoding="utf-8") as f:
    for h, m in zip(hodiny, mzdy):
        f.write(f"{h} h → {m} Kč\n")
print("Výsledek uložen do data/vykaz1.txt")
```
