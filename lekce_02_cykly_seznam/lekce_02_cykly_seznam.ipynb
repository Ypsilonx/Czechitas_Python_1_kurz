## 1. Cyklus `for` a `range()`

`for` prochází jeden prvek za druhým – seznam, řetězec, nebo číselný rozsah přes `range()`.

```python
# Průchod přes seznam a řetězec – for funguje na čemkoliv "iterovateiném"
barvy = ["červená", "zelená", "modrá"]
for barva in barvy:
    print(barva)

print()
for pismeno in "Python":
    print(pismeno)
```

```python
# range() – generuje číselný rozsah
# range(stop)           → 0 až stop-1
# range(start, stop)    → start až stop-1
# range(start, stop, krok)

print("range(5):", list(range(5)))
print("range(2, 8):", list(range(2, 8)))
print("range(0, 20, 3):", list(range(0, 20, 3)))

# Praktické použití – vypiš čísla dělitelná 3 v rozsahu 1–30
print("\nDělitelná 3:")
for i in range(1, 31):
    if i % 3 == 0:
        print(i, end=" ")
```

```python
# Cvičení – tombola: vylosuj 3 náhodná čísla
import random

pocet_listku = 1000
print("Výherní čísla:")
for i in range(1, 4):
    vyherni = random.randint(1, pocet_listku)
    print(f"  {i}. výherní los: {vyherni}")
```

## 2. Seznam (`list`) a `append()`

Seznam je seřazená kolekce – prvky mají index. Dá se za běhu rozšiřovat pomocí `.append()`.

```python
# Sestavení seznamu hostů za běhu programu
# (místo input() používám pevný seznam, aby notebook šel spustit bez zadávání)
jmena = ["Anna", "Petr", "Jana", "Marek", "konec"]

guest_list = []
for jmeno in jmena:
    if jmeno == "konec":
        break                   # break ukončí cyklus okamžitě
    guest_list.append(jmeno)

print("Hosté:", guest_list)
print("Počet hostů:", len(guest_list))
```

```python
# Filtrování ze seznamu slovníků – výběr dárku v rozsahu cen
list_of_items = [
    {"title": "Modrá kravata",           "price": 239,  "in_stock": True},
    {"title": "Luxusní psací pero",      "price": 1599, "in_stock": True},
    {"title": "Degustační balíček kávy", "price": 599,  "in_stock": True},
    {"title": "Parfém",                  "price": 559,  "in_stock": False},
    {"title": "Čajová konvička s hrnky", "price": 899,  "in_stock": True},
    {"title": "Sklenice na víno",        "price": 799,  "in_stock": True},
    {"title": "Fitness náramek",         "price": 2399, "in_stock": False},
]

# Chci první dostupný dárek v rozsahu 500–1000 Kč
for item in list_of_items:
    if 500 <= item["price"] <= 1000 and item["in_stock"]:
        print(f"Vybraný dárek: {item['title']} ({item['price']} Kč)")
        break
```

## 3. Akumulace výsledků do slovníku

Tahle technika mi přišla super – procházím seznam a průběžně stavím slovník  
s výsledky (součty, počty...). Klíč = osoba/kategorie, hodnota = akumulovaný výsledek.

```python
# Společné nákupy spolubydlících – kdo kolik utratil?
purchase_list = [
    {"person": "Petr",  "item": "Prací prášek",       "value": 399},
    {"person": "Ondra", "item": "Savo",                "value": 80},
    {"person": "Petr",  "item": "Toaletní papír",      "value": 65},
    {"person": "Libor", "item": "Pivo",                "value": 124},
    {"person": "Petr",  "item": "Pytel na odpadky",    "value": 75},
    {"person": "Míša",  "item": "Utěrky na nádobí",    "value": 130},
    {"person": "Ondra", "item": "Toaletní papír",      "value": 120},
    {"person": "Míša",  "item": "Pečící papír",        "value": 30},
    {"person": "Zuzka", "item": "Savo",                "value": 80},
    {"person": "Pavla", "item": "Máslo",               "value": 50},
    {"person": "Ondra", "item": "Káva",                "value": 300},
]

# Akumulace – pokud osoba už ve slovníku je, přičtu; jinak vytvořím nový záznam
sum_per_person = {}
for nakup in purchase_list:
    osoba = nakup["person"]
    if osoba in sum_per_person:
        sum_per_person[osoba] += nakup["value"]
    else:
        sum_per_person[osoba] = nakup["value"]

total = sum(sum_per_person.values())
prumer = total / len(sum_per_person)

print("Výdaje na osobu:")
for osoba, castka in sum_per_person.items():
    print(f"  {osoba}: {castka} Kč")
print(f"\nCelkem: {total} Kč | Průměr: {prumer:.0f} Kč/osoba")
```

```python
# Vyrovnání – kdo má dostat a kdo doplatit?
print("Vyrovnání:")
for osoba, castka in sum_per_person.items():
    rozdil = round(castka - prumer)
    if rozdil >= 0:
        print(f"  {osoba} má dostat zpět: {rozdil} Kč")
    else:
        print(f"  {osoba} má doplatit:   {-rozdil} Kč")
```

## 4. Cyklus `while`

`while` se opakuje **dokud je podmínka pravdivá** – hodí se tam,  
kde nevíme předem kolikrát se cyklus provede.

```python
# Hra: hádání tajného čísla
# (simuluji hráče automaticky, aby notebook šel spustit bez vstupu)
import random

tajne_cislo = random.randint(1, 20)
pokusy = [3, 15, 7, 12, tajne_cislo]  # simulované tipy hráče
print(f"(tajné číslo pro kontrolu: {tajne_cislo})\n")

for tip in pokusy:
    print(f"Hráč hádá: {tip}")
    if tip < tajne_cislo:
        print("  → příliš malé, zkus výš")
    elif tip > tajne_cislo:
        print("  → příliš velké, zkus níž")
    else:
        print("  → TREFIL! 🎉")
        break
```

## 5. Bonus – prvočísla

Číslo je prvočíslo, pokud ho nelze beze zbytku dělit žádným číslem kromě 1 a sebe sama.  
Stačí testovat dělitele od 2 do `n-1`.

```python
def je_prvocislo(n):
    if n <= 1:
        return False
    for delitel in range(2, n):
        if n % delitel == 0:
            return False   # našli jsme dělitele → není prvočíslo
    return True

# Test pro čísla 1–20
prvocisla = [n for n in range(1, 21) if je_prvocislo(n)]
print("Prvočísla do 20:", prvocisla)
```

Zadání: [ukol-02](https://github.com/andywaltlova/python-1-podzim-2022/blob/master/ukoly/ukol-02.md)

Projdi seznam států a vypiš všechny státy z hledaného regionu.  
Plný dataset je v souboru [data/staty.json](../data/staty.json).

```python
import json
from pathlib import Path

# Načti data ze sdíleného souboru
staty = json.loads(Path("../data/staty.json").read_text(encoding="utf-8"))

hledany_region = "Europe"  # změň na libovolný: Africa, Asia, Americas, Oceania...
nalezene = [s["name"] for s in staty if s.get("region") == hledany_region]

if nalezene:
    print(f"Státy v regionu '{hledany_region}' ({len(nalezene)}):")
    for stat in sorted(nalezene):
        print(f"  {stat}")
else:
    print("Neznámý region")
```
